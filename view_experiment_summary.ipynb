# Experiment Cache Summary

Load and inspect the CSV table produced by `summarize_experiment_cache.py`.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 0)
pd.set_option("display.max_colwidth", None)
```

```python
SUMMARY_PATH = Path("experiment_cache/cache_mean_summary.csv")

if not SUMMARY_PATH.exists():
    raise FileNotFoundError(
        f"Could not find {SUMMARY_PATH}. Run `python summarize_experiment_cache.py` first."
    )

df = pd.read_csv(SUMMARY_PATH)
print(f"Loaded {len(df)} rows from {SUMMARY_PATH}")
```

## Full Table

```python
display(df)
```

## Compact Result View

```python
columns = [
    "experiment",
    "pde_key",
    "split",
    "ood_filename",
    "method",
    "kernel",
    "pca",
    "n_eval",
    "mean_percent",
    "std_percent",
]

compact = df[[col for col in columns if col in df.columns]].copy()
compact = compact.sort_values(["experiment", "pde_key", "split", "pca", "method"])
display(compact)
```

## Pivot Table

```python
pivot = compact.pivot_table(
    index=["experiment", "pde_key", "split", "ood_filename"],
    columns=["method"],
    values="mean_percent",
    aggfunc="first",
)

display(pivot)
```

## Optional Filters

```python
# Edit these lists to focus the table. Leave a list empty to keep all values.
EXPERIMENTS = []  # e.g. ["framework1_operator"]
PDES = []         # e.g. ["conservation", "param_wave"]
SPLITS = []       # e.g. ["test", "ood_ood_par_h5"]
PCA = None        # True, False, or None

filtered = compact.copy()
if EXPERIMENTS:
    filtered = filtered[filtered["experiment"].isin(EXPERIMENTS)]
if PDES:
    filtered = filtered[filtered["pde_key"].isin(PDES)]
if SPLITS:
    filtered = filtered[filtered["split"].isin(SPLITS)]
if PCA is not None:
    filtered = filtered[filtered["pca"] == PCA]

display(filtered)
```
