```python
import h5py
import json

import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import Ridge, LinearRegression, SGDRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, RBF
import time
from sklearn.kernel_ridge import KernelRidge
from sklearn.model_selection import train_test_split
```

# Out-of-distribution: variance

```python
with h5py.File('ood_var.h5', 'r') as f:
    # List all groups/datasets
    dataset_var = f["data"]   # This does NOT load data
    dataset_var = dataset_var[:]
    print(dataset_var.shape)        # Fast
    coef_var = f["coeffs"]
    coef_var = coef_var[:]
    print(coef_var.shape)
```

```python
idx = 0
fig, ax = plt.subplots(1,3,figsize=(15,5))
ax[0].plot(dataset_var[idx,0,:,0], label='initial condition')
ax[0].legend()
ax[1].plot(dataset_var[idx,-1,:,0], label='solution')
ax[1].legend()
ax[2].plot(coef_var[idx], label='coef')
ax[2].legend()
```

```python
idx = 20
fig, ax = plt.subplots(1,3,figsize=(15,5))
ax[0].plot(dataset_var[idx,0,:,0], label='initial condition')
ax[0].legend()
ax[1].plot(dataset_var[idx,-1,:,0], label='solution')
ax[1].legend()
ax[2].plot(coef_var[idx], label='coef')
ax[2].legend()
```
