```python
import h5py
import json

import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import Ridge, LinearRegression, SGDRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, RBF
import time
from sklearn.kernel_ridge import KernelRidge
from sklearn.model_selection import train_test_split
```

## OOD: initial condition

```python
with h5py.File('ood_init.h5', 'r') as f:
    # List all groups/datasets
    dataset1 = f["data"]   # This does NOT load data
    dataset1 = dataset1[:]
    print(dataset1.shape)        # Fast
    coef1 = f["coeffs"]
    coef1 = coef1[:]
    print(coef1.shape)
```

```python
idx = 0
fig, ax = plt.subplots(1,3,figsize=(15,5))
ax[0].plot(dataset1[idx,0,:,0], label='initial')
ax[0].legend()
ax[1].plot(dataset1[idx,-1,:,0], label='sol')
ax[1].legend()
ax[2].plot(coef1[idx], label='parametric')
ax[2].legend()
```
