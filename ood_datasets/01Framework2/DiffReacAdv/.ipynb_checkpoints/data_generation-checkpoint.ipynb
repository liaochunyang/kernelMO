```python
import os
import numpy as np
import h5py
from tqdm import tqdm

import matplotlib.pyplot as plt
```

```python
from scipy.spatial.distance import cdist

def covariance(x,y, k_sigma, l):
    sq_norm = cdist(x, y, metric='sqeuclidean')
    return np.exp(-0.5 * sq_norm / l**2) * k_sigma**2
    

def Gaussian_Process(xc, nu, k_sigma, l):
    
    nx = len(xc)
    # generate parametric function
    K = covariance(xc.reshape(-1,1),xc.reshape(-1,1),k_sigma, l)
    alpha = np.random.multivariate_normal(mean=np.zeros(nx), cov=K)
    
    # periodic
    slope = (alpha[-1] - alpha[0]) / (xc[-1] - xc[0])
    alpha = alpha - np.outer(slope,xc).reshape(-1)
    
    # 
    min_val = np.min(alpha)
    max_val = np.max(alpha)
    
    alpha = (alpha - min_val) / (max_val - min_val)

    return nu * alpha
```

```python
def generate_initial_conditions(xc, n_samples, rng, amp_low=0, amp_up=1, k_tot=4, n_modes=2, mode='sin', nu=1, k_sigma=1, l=1):
    """Generate random sinusoidal initial conditions, normalized to [0, 1].
    Mimics the init_multi function from the original codebase."""
    nx = len(xc)
    x_len = xc[-1] - xc[0]
    u_all = np.zeros((n_samples, nx), dtype=np.float32)
    
    if mode == 'sin':
        for i in range(n_samples):
            # Select random wavenumber modes
            selected_k = rng.choice(k_tot, size=n_modes, replace=False) + 1
            u = np.zeros(nx, dtype=np.float64)
            for k in selected_k:
                amp = rng.uniform(amp_low, amp_up)
                phase = rng.uniform(0, 2 * np.pi)
                wavenumber = 2 * np.pi * k / x_len
                u += amp * np.sin(wavenumber * xc + phase)
            # Random sign flip
            if rng.random() < 0.5:
                u = -u
            # Remove linear trend for periodicity
            slope = (u[-1] - u[0]) / x_len
            u -= slope * xc
            # Normalize to [0, 1]
            u_min, u_max = u.min(), u.max()
            if u_max - u_min > 1e-10:
                u = (u - u_min) / (u_max - u_min)
            u_all[i] = u.astype(np.float32)
            
            
    elif mode == 'GP':
        
        for i in range(n_samples):
            u_all[i] = Gaussian_Process(xc, nu, k_sigma, l)
           
    else:
        raise ValueError('Wrong initial condition mode')
        
    return u_all
```

# Diffusion-Reaction-Advection Equation

```python
def solve_diffreactadv(u0, nu, v, rho, p, q, x_range, t_range, nx, t_num, CFL=0.4, mode="periodic"):
    """
    Solve: u_t + v u_x = nu u_xx + rho u^p (1 - u^q)

    Returns: solution array of shape (t_num, nx)
    """

    dx = x_range / nx
    dx_inv = 1.0 / dx
    dt_save = t_range / (t_num - 1)

    # ------------------------
    # Boundary conditions
    # ------------------------
    def bc(u):
        _u = np.zeros(nx + 4)
        _u[2:nx+2] = u

        if mode == "periodic":
            _u[0:2] = u[-2:]
            _u[nx+2:nx+4] = u[0:2]
        else:
            _u[0:2] = u[0]
            _u[nx+2:nx+4] = u[-1]

        return _u

    # ------------------------
    # Limiter (same as yours)
    # ------------------------
    def vl_limiter(a, b, c, alpha=2.0):
        return np.sign(c) * (0.5 + 0.5 * np.sign(a * b)) * \
               np.minimum(alpha * np.minimum(np.abs(a), np.abs(b)), np.abs(c))

    def limiting(_u):
        duL = _u[1:nx+3] - _u[0:nx+2]
        duR = _u[2:nx+4] - _u[1:nx+3]
        duM = 0.5 * (_u[2:nx+4] - _u[0:nx+2])

        gradu = vl_limiter(duL, duR, duM)

        uL = np.zeros_like(_u)
        uR = np.zeros_like(_u)

        uL[1:nx+3] = _u[1:nx+3] - 0.5 * gradu
        uR[1:nx+3] = _u[1:nx+3] + 0.5 * gradu

        return uL, uR

    # ------------------------
    # Flux (advection + diffusion)
    # ------------------------
    def compute_flux(u):
        _u = bc(u)
        uL, uR = limiting(_u)

        # Advection flux (Rusanov)
        fL = v * uL
        fR = v * uR

        a = np.abs(v)  # constant speed

        f_adv = 0.5 * (
            fR[1:nx+2] + fL[2:nx+3]
            - a * (uL[2:nx+3] - uR[1:nx+2])
        )

        # Diffusion (central)
        f_diff = -nu * (_u[2:nx+3] - _u[1:nx+2]) * dx_inv

        return f_adv + f_diff

    # ------------------------
    # Reaction term
    # ------------------------
    def reaction(u, dt):
        if p == 1 and q == 1:
            # exact logistic update
            return 1.0 / (1.0 + np.exp(-rho * dt) * (1.0 - u) / u)
        else:
            # explicit Euler (substeps for stability)
            for _ in range(5):
                u = u + (rho * u**p * (1 - u**q)) * (dt / 5.0)
            return u

    # ------------------------
    # Update step (RK2 + splitting)
    # ------------------------
    def update(u, u_tmp, dt):
        # reaction (Strang splitting idea)
        u = reaction(u, dt)

        f = compute_flux(u_tmp)
        u = u - dt * dx_inv * (f[1:nx+1] - f[0:nx])

        return u

    # ------------------------
    # Time integration
    # ------------------------
    u = u0.copy()
    uu = np.zeros((t_num, nx), dtype=np.float32)
    uu[0] = u

    t = 0.0
    i_save = 1
    t_save = dt_save

    while i_save < t_num:

        # CFL conditions
        dt_adv = CFL * dx / (abs(v) + 1e-8)
        dt_dif = CFL * 0.5 * dx**2 / (nu + 1e-8)
        dt_reac = CFL / (rho + 1e-8)

        dt = min(dt_adv, dt_dif, dt_reac, t_save - t)

        if dt < 1e-12:
            uu[i_save:] = u
            break

        # RK2
        u_tmp = update(u, u, dt * 0.5)
        u = update(u, u_tmp, dt)

        t += dt

        if t >= t_save - 1e-10:
            uu[i_save] = u.astype(np.float32)
            i_save += 1
            t_save = i_save * dt_save

    return uu
```

```python
def generate_diffusion_reaction_advection_dataset(data_dir, n_param_sets, ics_per_param,
                                   nu_center, v_center, rho_center, gamma,
                                   x_range, t_range, nx, t_num, file_name, 
                                   amp_low = 0, amp_up=1, k_tot=4, n_modes=2,
                                   mode='sin', nu=1, k_sigma=1, l=1, seed=42):
    """
    Generate diffusion reaction advection dataset and save to HDF5.
    Saves:
      - data_dir/solutions.h5: 'data' (N, t_num, nx, 1), 'coeffs' (N, 5)
    """
    os.makedirs(data_dir, exist_ok=True)
    h5_path = os.path.join(data_dir, file_name)

    rng = np.random.default_rng(seed)
    dx = x_range / nx
    xc = np.linspace(0, x_range, nx + 1)[:-1] + 0.5 * dx

    total = n_param_sets * ics_per_param
    all_solutions = np.zeros((total, t_num, nx, 1), dtype=np.float32)
    all_coeffs = np.zeros((total, 5), dtype=np.float32)
    idx = 0
    failed = 0

    for p in tqdm(range(n_param_sets), desc="Generating data"):
        # Sample PDE coefficients
        # u_t + v u_x = nu u_xx + rho u^p (1 - u^q)
        
        # nu (alpha1)
        nu_half = abs(nu_center) * gamma
        alpha1 = rng.uniform(nu_center - nu_half, nu_center + nu_half)
        # v (alpha2)
        v_half = abs(v_center) * gamma
        alpha2 = rng.uniform(v_center - v_half, v_center + v_half)
        # rho (alpha3)
        rho_half = abs(rho_center) * gamma
        alpha3 = rng.uniform(rho_center - rho_half, rho_center + rho_half)
        # alpha4, alpha5
        k = rng.uniform(1, 3, size=2)
        coeff = np.array([alpha1, alpha2, alpha3, k[0], k[1]], dtype=np.float32)

        # Generate ICs for this parameter set
        ics = generate_initial_conditions(xc, ics_per_param * 2, rng, amp_low = amp_low, amp_up = amp_up,
                                         k_tot = k_tot, n_modes = n_modes, mode = mode, 
                                          nu = nu, k_sigma = k_sigma, l=l)  # generate extra in case some fail
        count = 0
        for ic_idx in range(len(ics)):
            if count >= ics_per_param:
                break
            try:
                sol = solve_diffreactadv(ics[ic_idx].astype(np.float64),
                                        alpha1, alpha2, alpha3, k[0], k[1],
                                        x_range, t_range, nx, t_num, CFL=0.4)
                norm = np.linalg.norm(sol)
                if norm < 2000 and norm > 1 and np.all(np.isfinite(sol)):
                    all_solutions[idx, :, :, 0] = sol
                    all_coeffs[idx] = coeff
                    idx += 1
                    count += 1
                else:
                    failed += 1
            except Exception:
                failed += 1


    # Trim to actual size
    all_solutions = all_solutions[:idx]
    all_coeffs = all_coeffs[:idx]

    # Save
    with h5py.File(h5_path, "w") as hf:
        hf.create_dataset("data", data=all_solutions)
        hf.create_dataset("coeffs", data=all_coeffs)

    print(f"Saved {idx} samples to {h5_path}")
    return h5_path
```

# Training Dataset

```python
# ============ Data / PDE config ============
DATA_DIR        = "dataset_simple/"   # where to save/load generated data
T_NUM           = 64                  # timesteps
X_NUM           = 128                 # spatial grid points
X_NUM_MODEL     = 64                  # subsampled spatial points for model
T_RANGE         = (0.0, 2.0)
X_RANGE         = (0.0, 2.0)
COEFF_NUM       = 5                   # [epsilon, k1, k2, k3]
INPUT_LEN       = 1                   # input timesteps
INPUT_STEP      = 1
OUTPUT_STEP     = 2
NORMALIZE       = True

# Generation parameters
N_PARAM_SETS    = 280                 # number of different PDE coefficient sets
ICS_PER_PARAM   = 50                  # initial conditions per coefficient set
TOTAL_SAMPLES   = N_PARAM_SETS * ICS_PER_PARAM  # 2000 total

# PDE coefficient ranges 
NU_CENTER       = 0.01                # diffusion coefficient center
V_CENTER        = - 1.0
RHO_CENTER      = 1.0                 # flux coefficient center
PARAM_GAMMA     = 0.1                 # +/- 10% range

# Initial condition: default setting
AMP_LOW     = 0
AMP_UP      = 1 
K_TOT       = 4
N_MODES     = 2
MODE        = 'sin'
NU          = 1
K_SIGMA     = 1
L           = 1
# Random Seed
SEED        = 42 
```

```python
file_name = "solutions.h5"
# Generate or load data
h5_path = os.path.join(DATA_DIR, file_name)

if os.path.exists(h5_path):
    print(f"Data already exists at {h5_path}, loading...")
    with h5py.File(h5_path, "r") as hf:
        print(f"  Solutions shape: {hf['data'].shape}")
        print(f"  Coefficients shape: {hf['coeffs'].shape}")
else:
    print("Generating conservation law dataset...")
    generate_diffusion_reaction_advection_dataset(
        DATA_DIR, N_PARAM_SETS, ICS_PER_PARAM,
        NU_CENTER, V_CENTER, RHO_CENTER, PARAM_GAMMA,
        X_RANGE[1], T_RANGE[1], X_NUM, T_NUM, file_name, 
        amp_low = AMP_LOW, amp_up=AMP_UP, k_tot=K_TOT, n_modes=N_MODES,mode=MODE, 
        nu=NU, k_sigma=K_SIGMA, l=L, seed=SEED)
```
